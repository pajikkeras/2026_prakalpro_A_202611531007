### 2.2.2 Membuat Repository GitHub

Setelah program dibuat dan diuji pada VS Code, proyek dapat dihubungkan dengan GitHub. Repository digunakan untuk menyimpan kode program secara daring.

Tahapan yang dilakukan yaitu:

1. Buat repository baru di GitHub.
2. Buka folder proyek lewat VS Code.
3. Aktifkan terminal yang ada di dalam VS Code.
4. Inisialisasi Git pada folder proyek tersebut.
5. Tambahkan file ke dalam staging area.
6. Commit untuk mencatat perubahan.
7. Hubungkan repository lokal ke repository GitHub.
8. Unggah kode ke GitHub dengan perintah push.


```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

Penjelasan:

1. "git init" <br>
Perintah ini dipakai buat "menyalakan" Git di folder proyek kita. Jadi folder yang tadinya biasa saja, sekarang jadi bisa dipantau sama Git — artinya setiap perubahan yang kita buat di file-file dalam folder itu bisa mulai dicatat.<br>

2. "git add" . <br>
Perintah ini dipakai buat "mengumpulkan" semua file yang ada di folder proyek, biar siap disimpan. Tanda titik (.) artinya "ambil semua", jadi kita nggak perlu sebut file satu-satu, cukup satu perintah aja buat ambil semuanya sekaligus.<br>

3. "git commit -m "Initial commit" <br>
Perintah ini dipakai buat benar-benar menyimpan perubahan yang tadi sudah dikumpulkan. Bagian -m "Initial commit" itu semacam catatan singkat yang kita tulis, buat kasih tahu "ini simpanan tentang apa". Biasanya kalau ini penyimpanan pertama kali, orang suka nulis "Initial commit" yang artinya kurang lebih "simpanan awal".<br>

4. "git branch -M main"<br>
Perintah ini dipakai buat kasih nama main ke jalur kerja (branch) yang lagi kita pakai. Dulu biasanya nama defaultnya master, tapi sekarang kebanyakan orang (termasuk GitHub) pakai nama main. Jadi perintah ini semacam "ganti nama jalur kerja kita jadi main".<br>

5. "git remote add origin https://github.com/username/nama-repository.git"<br>
Perintah ini dipakai buat menyambungkan folder proyek yang ada di laptop kita dengan repository yang ada di GitHub. Kata origin itu cuma nama panggilan singkat buat alamat GitHub yang panjang tadi, biar ke depannya kita nggak perlu ketik link lengkapnya terus-terusan.<br>

6. "git push -u origin main"<br>
Perintah ini dipakai buat benar-benar mengirim (upload) semua yang sudah kita simpan tadi dari laptop ke GitHub, tepatnya ke jalur main. Bagian -u itu semacam "diingat" oleh Git, jadi lain kali kalau mau upload lagi, kita cukup ketik git push aja tanpa perlu tulis origin main lagi.<br>
