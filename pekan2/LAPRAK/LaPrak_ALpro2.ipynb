## 2.1 Dasar Teori
### 2.1.1 Variabel
Variabel merupakan lokasi penyimpanan yang digunakan untuk menyimpan suatu nilai atau data dalam program, dan ketika sebuah variabel didefinisikan maka Python secara otomatis mengalokasikan ruang di memori untuk menyimpan nilai tersebut. Berbeda dengan sejumlah bahasa pemrograman lain, Python tidak mengharuskan programmer mendeklarasikan tipe data secara eksplisit saat membuat variabel; cukup dengan menuliskan nama variabel beserta nilainya, dan Python akan menentukan tipe datanya secara otomatis. Pendekatan ini dikenal dengan istilah dynamically typed, yang berarti tipe data suatu variabel dapat berubah secara dinamis selama proses eksekusi program berlangsung.

### 2.1.2 Numeric
Python secara otomatis menentukan tipe data suatu variabel berdasarkan nilai yang diberikan kepadanya. Tipe data numerik pada Python terdiri atas tipe int, yaitu tipe data untuk menyimpan bilangan bulat tanpa nilai desimal, tipe float untuk menyimpan bilangan berkoma (floating point), serta tipe complex untuk menyimpan bilangan kompleks dengan format a + bj, di mana a merupakan bagian real dan b merupakan bagian imajiner . Secara garis besar, jenis tipe data angka pada Python terbagi menjadi integer (bilangan bulat) dan float (bilangan pecahan).

### 2.1.3 String
String atau str adalah tipe data yang digunakan untuk menyimpan nilai dalam bentuk teks, yang penulisannya dapat diapit menggunakan tanda kutip ganda maupun tanda kutip tunggal. Dalam praktiknya, data string sering dikombinasikan dengan variabel dari tipe data lain, dan terdapat beberapa cara untuk memformat hasil kombinasi tersebut, yaitu menggunakan operator %, fungsi format(), maupun f-string, di mana f-string saat ini lebih disarankan karena sintaksnya lebih ringkas dan mudah dipahami.

### 2.1.4 Boolean
Tipe data boolean merupakan tipe data yang hanya memiliki nilai benar atau salah, dan pada Python jenis tipe data ini termasuk ke dalam kategori tipe data dasar selain tipe data angka dan tipe data teks. Contoh penerapannya dapat dilihat pada deklarasi variabel dengan nilai True atau False, sebagaimana pada tipe data lain yang jenisnya ditentukan otomatis oleh Python berdasarkan nilai yang diberikan.

### 2.1.5 Konstanta
Dalam bahasa pemrograman Python, penulisan konstanta pada dasarnya mengikuti aturan penulisan variabel pada umumnya, namun secara konvensi nilai yang dimaksudkan sebagai konstanta dituliskan menggunakan huruf kapital untuk menandakan bahwa nilainya tidak dimaksudkan untuk diubah selama program berjalan. Salah satu contoh nya adalah nilai PI

### 2.1.6 Input
Fungsi input() merupakan fungsi bawaan (built-in) pada Python yang digunakan untuk menerima masukan dari pengguna melalui keyboard, dan fungsi ini bekerja dengan menerima satu parameter opsional berupa teks yang akan ditampilkan sebagai prompt sebelum pengguna memasukkan data. Nilai yang dimasukkan pengguna melalui fungsi input() akan selalu dikembalikan dalam bentuk tipe data string, sehingga apabila nilai tersebut akan digunakan dalam perhitungan matematis, perlu dilakukan konversi terlebih dahulu ke tipe data numerik menggunakan fungsi seperti int() atau float(). Pada versi Python 3, seluruh proses pengambilan input dari keyboard cukup dilakukan menggunakan fungsi input(), karena fungsi raw_input() yang terdapat pada Python 2 telah digabungkan ke dalamnya.

### 2.2.4 Program Input "A_2611531007"

### 2.2.5 Program Konstanta "A_2611531007"

```python
# Buat file dengan nama Konstanta_1234
# Programini menggunakan konstanta untuk menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14
print("pi: %F" % (PI))
jari_1007 = float(input('Masukkan nilai jari jari: '))
luas_1007 = PI * jari_1007 * jari_1007
print("Luas Lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_1007, luas_1007))
```

Penjelasan Program:

1. Baris 1-3 merupakan baris komentar yang membarikan intruksi tentang program .
2. Baris 4 Mengimpor penanda tipe (type hint) Final dari modul bawaan typing untuk menegaskan bahwa suatu variabel ditujukan sebagai konstanta (nilainya tidak boleh diubah).
3. Baris 5 Mendeklarasikan konstanta bernama PI bernilai 3.14 dan menandainya menggunakan : Final agar diperlakukan sebagai nilai tetap.
4. Baris 6 Mencetak nilai konstanta PI menggunakan format penulisan string legacy (%f), yang secara default akan menampilkan 6 angka di belakang koma (3.140000). Meminta masukan (input) dari pengguna untuk nilai jari-jari, lalu langsung mengonversi teks masukan tersebut menjadi tipe bilangan desimal menggunakan fungsi float(), dan menyimpannya ke variabel jari_1007.
5. Baris 7 Menghitung luas lingkaran dengan rumus 
 (PI * jari_1007 * jari_1007) dan menyimpannya ke dalam variabel luas_1007.
6. Baris 8 Mencetak hasil ke layar menggunakan format penulisan string (%.2f), di mana %.2f membatasi tampilan angka desimal variabel jari_1007 dan luas_2007 agar hanya memunculkan 2 angka di belakang koma.

Berdasarkan hasil praktikum yang telah dilakukan mengenai pengenalan variabel dan tipe data pada bahasa pemrograman Python, dapat disimpulkan bahwa variabel merupakan tempat penyimpanan nilai atau data yang dapat digunakan dan diubah selama program berjalan, tanpa perlu dideklarasikan tipe datanya secara eksplisit, karena Python bersifat dynamically typed sehingga tipe data suatu variabel ditentukan secara otomatis berdasarkan nilai yang diberikan kepadanya.
