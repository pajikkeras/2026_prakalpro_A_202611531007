## 2.1 Dasar Teori

Struktur kontrol percabangan (*selection/decision structure*) merupakan salah satu dari empat struktur kontrol dasar dalam pemrograman selain runtutan (*sequence*), pengulangan (*loop*), dan subprogram (*procedure*). Struktur kontrol pemilihan digunakan untuk memilih tindakan yang akan dijalankan berdasarkan suatu kondisi tertentu, dengan beberapa jenis pemilihan seperti If, If-Else, If bersarang, dan Switch/Case. Secara umum, struktur percabangan digunakan untuk memecahkan masalah yang memiliki banyak pernyataan dan mengambil keputusan di antara pernyataan-pernyataan tersebut, sehingga program dapat menyeleksi satu pernyataan dari pernyataan lain dalam menyelesaikan suatu permasalahan.

Penelitian di bidang *computing education* menunjukkan bahwa pernyataan kondisional merupakan salah satu konsep pemrograman yang paling menentukan keberhasilan pemrograman pemula, sekaligus salah satu yang paling banyak menimbulkan kesalahpahaman. Ebrahimi (1994) mengelompokkan kesalahan pemrogram pemula (*novice programmer errors*) ke dalam dua kategori besar, yaitu kesalahan pada konstruksi bahasa (*language constructs*) dan kesalahan pada penyusunan rencana program (*plan composition*), di mana struktur percabangan termasuk salah satu konstruksi yang paling sering menjadi sumber kesalahan tersebut. Sejalan dengan itu, Green (1977) dalam studinya terhadap keterbacaan (*comprehensibility*) pernyataan kondisional pada pemrogram profesional menemukan bahwa bentuk penulisan suatu kondisi—termasuk penggunaan negasi dan penyusunan cabang alternatif—memengaruhi seberapa cepat dan seberapa akurat seseorang dapat memahami maksud dari suatu program. Temuan-temuan ini menjadi dasar penting mengapa pemahaman menyeluruh terhadap berbagai varian struktur percabangan, mulai dari yang paling sederhana hingga yang paling kompleks, perlu dikuasai sejak tahap awal pembelajaran pemrograman.

### 2.1.1 Pernyataan IF

Pernyataan `if` adalah bentuk percabangan paling dasar dan paling sederhana. Percabangan if melakukan pengecekan kondisi dan menentukan arah jalannya program, di mana hanya akan menghasilkan dua kemungkinan hasil, yaitu benar (*True*) atau salah (*False*). Jika kondisi bernilai benar, maka blok pernyataan di bawahnya akan dieksekusi; jika kondisi bernilai salah, blok tersebut akan dilewati begitu saja tanpa ada aksi alternatif. Struktur ini disebut sebagai bentuk paling dasar karena hanya melibatkan satu jalur keputusan tanpa cabang lanjutan, sehingga aliran program setelah blok `if` akan tetap berlanjut ke baris berikutnya, baik kondisi tersebut terpenuhi maupun tidak.

Karena sifatnya yang hanya memberi satu jalur eksekusi, struktur `if` tunggal umumnya digunakan untuk kasus-kasus validasi atau pengecekan sederhana yang tidak membutuhkan tindakan balasan ketika kondisi tidak terpenuhi—misalnya sekadar mencetak peringatan tanpa perlu menghentikan atau mengalihkan jalannya program. Meskipun struktur `if` tunggal tampak sederhana secara sintaksis, Kwon (2017) dalam penelitiannya terhadap kesalahpahaman pemrogram pemula menemukan bahwa banyak pemula justru salah dalam menyusun rencana pemecahan masalah (*problem-solving plan*) sejak tahap penentuan kondisi awal, misalnya keliru dalam merumuskan ekspresi boolean yang seharusnya diuji atau salah menentukan batas nilai suatu perbandingan. Akibatnya, meskipun sintaks `if` yang ditulis benar secara tata bahasa (tanda titik dua dan indentasi terpenuhi), logika program tetap dapat keliru karena kondisi yang diuji tidak mencerminkan maksud sebenarnya dari permasalahan yang ingin diselesaikan. Hal ini menegaskan bahwa penguasaan `if` tunggal bukan hanya soal kebenaran sintaks, melainkan juga soal kemampuan merumuskan ekspresi logis yang tepat terhadap konteks permasalahan yang dihadapi, sebelum struktur percabangan yang lebih kompleks dapat dipahami dengan baik.

### 2.1.2 Nested IF

*Nested if* adalah penempatan satu struktur `if` di dalam struktur `if` yang lain. Secara sederhana, IF bersarang atau nested IF adalah penggunaan struktur IF di dalam IF, dan kondisi seperti ini sering digunakan untuk kode program yang sudah cukup kompleks. Pada struktur ini, blok kode di dalam `if` bagian dalam (*inner if*) hanya akan dieksekusi apabila kondisi pada `if` bagian luar (*outer if*) juga terpenuhi; apabila kondisi pada if luar tidak terpenuhi, maka program akan langsung melompat ke bagian else tanpa mengevaluasi kondisi bagian dalam sama sekali. Dengan kata lain, kondisi pada if bagian dalam bersifat bersyarat ganda—ia hanya relevan dan hanya diperiksa jika prasyarat pada if bagian luar sudah terlebih dahulu dipenuhi.

Struktur nested if pada dasarnya digunakan ketika suatu keputusan bergantung pada lebih dari satu tingkat pertimbangan yang bersifat hierarkis, misalnya pertimbangan pertama menentukan kategori umum suatu kasus, sementara pertimbangan kedua di dalamnya menentukan rincian lebih lanjut dari kategori tersebut. Namun demikian, Gobil, Shukor, dan Mohtar (2009) dalam penelitiannya terhadap kesulitan pemrogram pemula pada struktur seleksi (*selection structure*) menemukan bahwa semakin dalam tingkat penyarangan (*nesting level*) suatu percabangan, semakin besar pula kesulitan mahasiswa dalam melacak kondisi mana yang sedang aktif dieksekusi pada suatu titik program, terutama ketika mereka diminta melakukan penelusuran kode (*code tracing*) secara manual tanpa bantuan alat bantu visual. Izu, Mirolo, dan Weerasinghe (2018) memperkuat temuan tersebut melalui studi mengenai penalaran pemrogram pemula terhadap pembalikan pernyataan kondisional (*reversing conditional statements*), yang menunjukkan bahwa kesalahan penalaran semakin meningkat seiring bertambahnya jumlah cabang dan tingkat kedalaman struktur kondisional yang harus dipahami secara simultan, karena pemrogram pemula harus menahan (*hold*) lebih dari satu kondisi sekaligus di dalam ingatan kerja mereka saat menelusuri alur program. Oleh karena itu, meskipun suatu bahasa pemrograman tidak membatasi jumlah tingkat penyarangan `if` secara teknis, penggunaan nested if yang berlebihan pada kode program yang panjang sebaiknya dihindari atau digantikan dengan struktur lain yang lebih terstruktur, seperti `if-elif-else` dengan operator logika, agar keterbacaan dan kebenaran logika program tetap terjaga, sekaligus mengurangi beban kognitif bagi siapa pun yang membaca atau memelihara kode tersebut di kemudian hari.

### 2.1.3 Multi IF

Multi IF adalah struktur percabangan yang terdiri dari beberapa pernyataan `if` yang berdiri sendiri-sendiri (independen), dituliskan berurutan satu demi satu, tanpa menggunakan `elif` atau `else` untuk menghubungkannya. Karena setiap `if` dievaluasi secara terpisah dan tidak saling terkait satu sama lain, sifat khas dari Multi IF adalah kemungkinan lebih dari satu blok kode dieksekusi sekaligus apabila lebih dari satu kondisi bernilai benar (*True*) secara bersamaan—berbeda secara mendasar dengan `if-elif-else` yang hanya akan mengeksekusi satu blok saja meskipun beberapa kondisi sebenarnya sama-sama terpenuhi, karena begitu satu kondisi cocok, pengujian kondisi-kondisi berikutnya akan langsung dihentikan.

Karakteristik akumulatif inilah yang menjadikan Multi IF sangat relevan digunakan pada permasalahan di mana beberapa syarat bersifat tidak saling eksklusif (*non-mutually exclusive*), sehingga setiap syarat yang terpenuhi tetap perlu diberi perlakuan tersendiri tanpa mengabaikan syarat-syarat lain yang mungkin juga terpenuhi secara bersamaan—misalnya dalam perhitungan potongan harga bertingkat yang dapat ditumpuk (*stackable discount*) berdasarkan beberapa kriteria berbeda sekaligus, atau dalam pemberian beberapa label/kategori kepada satu data yang sama secara bersamaan. Naude, Denny, dan Luxton-Reilly (2024) dalam kajiannya mengenai pola-pola keliru (*anti-patterns*) pada pernyataan kondisional mahasiswa turut menyoroti bahwa kekeliruan penggunaan struktur kondisional sering terjadi ketika pemrogram pemula secara tidak sadar menggunakan `elif` pada situasi yang sebenarnya membutuhkan evaluasi kondisi independen (Multi IF), sehingga logika akumulasi yang diharapkan justru tidak pernah tercapai karena program berhenti pada kondisi pertama yang cocok dan mengabaikan kondisi-kondisi lain yang sebenarnya juga relevan. Dengan demikian, pemilihan antara Multi IF dan `if-elif-else` bukan sekadar pilihan gaya penulisan, melainkan keputusan yang secara langsung memengaruhi kebenaran logika dan hasil akhir dari suatu program.

### 2.1.4 IF Else

IF-ELSE adalah struktur kontrol percabangan yang digunakan untuk memilih di antara dua kemungkinan tindakan, tergantung pada apakah suatu kondisi bernilai benar (*True*) atau salah (*False*). Berbeda dengan `if` tunggal yang bisa "tidak melakukan apa-apa" jika kondisi salah, struktur `if-else` selalu menjalankan salah satu dari dua blok kode, sehingga program senantiasa memiliki keluaran atau tindakan tertentu apa pun hasil evaluasi kondisinya, tanpa pernah meninggalkan celah di mana tidak ada aksi yang diambil sama sekali.

Struktur ini sangat cocok digunakan ketika suatu permasalahan memiliki dua kemungkinan hasil yang saling eksklusif dan saling melengkapi, artinya jika salah satu kemungkinan tidak terjadi maka kemungkinan yang lain pasti terjadi, sehingga tidak ada ruang bagi hasil ketiga. Green (1977) menyatakan bahwa keberadaan cabang alternatif (`else`) yang eksplisit umumnya justru mempermudah pemrogram dalam memahami maksud keseluruhan program dibandingkan dengan struktur `if` tunggal yang berdiri sendiri tanpa alternatif, karena pembaca kode tidak perlu menduga-duga apa yang terjadi jika kondisi tidak terpenuhi—segala kemungkinan sudah tercakup secara eksplisit dalam kedua blok yang tersedia. Meski demikian, keberadaan dua jalur ini bukan berarti struktur if-else bebas dari potensi kesalahpahaman; Izu et al. (2018) menemukan bahwa pemrogram pemula tetap sering keliru ketika diminta membalik logika suatu kondisi `if-else` menjadi bentuk yang setara (misalnya mengubah kondisi dan menukar isi blok `if` dengan `else`), yang menunjukkan bahwa pemahaman terhadap ekuivalensi logis antara suatu kondisi dengan negasinya masih menjadi tantangan tersendiri bagi pemrogram pemula, meskipun sintaks if-else itu sendiri tergolong sederhana untuk dipelajari.

### 2.1.5 Match Case

Program akan mengambil nilai dari sebuah variabel (pada instruksi `match`), kemudian membandingkannya secara berurutan dengan nilai-nilai yang ada pada setiap blok `case`. Jika tidak ada `case` yang cocok, program dapat mengeksekusi blok *default* yang ditandai dengan simbol garis bawah (`case _:`), sehingga program tetap memiliki penanganan yang jelas untuk nilai apa pun yang tidak terdaftar secara eksplisit dalam daftar `case` yang tersedia.

Struktur `match-case` merupakan fitur pencocokan pola terstruktur (*structural pattern matching*) yang secara resmi diperkenalkan pada Python 3.10 melalui PEP 634 sebagai spesifikasi teknisnya (van Rossum & Bucher, 2021), dengan PEP 635 (Kohn & van Rossum, 2021) menjelaskan motivasi dan alasan (*rationale*) di balik penambahan fitur ini ke dalam bahasa Python. Menurut PEP 635, motivasi utama penambahan `match-case` adalah menyediakan alternatif yang lebih deklaratif dan lebih mudah dibaca dibandingkan rantai `if-elif` yang panjang, khususnya ketika jumlah kondisi yang diperiksa terhadap satu variabel yang sama semakin banyak, sehingga struktur kode menjadi lebih ringkas tanpa mengorbankan kejelasan maksud program. Berbeda dengan `switch-case` klasik pada bahasa pemrograman lain yang umumnya hanya membandingkan nilai secara langsung, `match-case` pada Python juga dirancang mampu mendekonstruksi (*destructure*) struktur data seperti *sequence*, *dictionary*, maupun objek instance kelas, sekaligus menangkap nilai-nilai di dalamnya sebagai bagian dari proses pencocokan pola itu sendiri, sehingga jangkauan kegunaannya jauh lebih luas daripada sekadar pengganti `if-elif` biasa untuk perbandingan nilai tunggal.

Meskipun tergolong fitur yang relatif baru dan belum banyak dikaji secara mendalam dalam literatur pendidikan pemrograman dibandingkan struktur percabangan klasik lainnya, prinsip dasar yang mendasarinya tetap sejalan dengan prinsip umum struktur seleksi: setiap kemungkinan nilai perlu ditangani secara eksplisit agar tidak ada kondisi yang luput dari penanganan program. Salah satu kesalahan umum dalam penerapan `match-case` adalah lupa menyertakan pola wildcard `case _` sebagai kondisi *default*, sebagaimana peran `else` pada struktur if-else, yang jika diabaikan dapat menyebabkan program tidak memberikan keluaran apa pun—atau bahkan menimbulkan galat (*error*)—ketika nilai yang diuji tidak cocok dengan pola manapun yang telah didefinisikan sebelumnya.

```python
# Buat file dengan nama match_castel_NIM.py
# Buat program untuk match castel
# Nama Variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input ()
# Program konversi nilai angka menjadi nama bulan

bulan_1007 = int(input("Masukkan angka bulan (1 - 12)=  "))

match bulan_1007:
    case 1:
        print("Januari")
    case 2 :
        print("Februari")
    case 3 :
        print("Maret")
    case 4 :
        print("April")
    case 5 :
        print("Mei")
    case 6 :
        print("Juni")
    case 7:
        print("juli")
    case 8 :
        print("Agustus")
    case 9 :
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12 :
        print("Desember")
    case _:
        print("Angka tidak valid")
```

### 2.2.6 Program Multi IF 2 "A_2611531007"

```python
# Buat file dengan nama multi_if2_NIM.py
# Buat program untuk kondisional if 
# Nama variabel ditambah 4 digit nim terakhir, contoh: total_belanja_1234
# Program ini menggunakan fungsi input ()
# Program ini menghitung diskon belanja

# Input dari user
total_belanja_1007 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_1007 = input ("Apakah Anda Member? (y/t):").strip().lower
is_member_1007 = input_member_1007 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_1007 = input("Apakah kode promo calid? (y/t):").strip().lower
kode_promo_valid_1007 = input_promo_1007 in ["y", "ya"]

total_diskon_persen_1007 = 0

# Multi if terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus
if total_belanja_1007 > 1000000:
    total_diskon_persen_1007 += 10 # Diskon belanjan besar

if is_member_1007:
    total_diskon_persen_1007 += 5 # Diskon member

if kode_promo_valid_1007:
    total_diskon_persen_1007 += 15 # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_1007 = total_belanja_1007 * (total_diskon_persen_1007 / 100)
total_bayar_1007 = total_belanja_1007 - nominal_diskon_1007

# Output Hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_1007}% (Rp {nominal_diskon_1007:,.0f})")
print(f"Total Bayar   : Rp {total_bayar_1007:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_1007}%")
# Output: Total diskon yang Anda dapatkan : 30% jika belanja > 1 juta , member, kode promo valid
```

# <h1 align = "center">KESIMPULAN <h1> 

Pernyataan if tunggal terbukti berguna untuk mengeksekusi satu blok kode hanya jika kondisi bernilai True, tanpa menyediakan tindakan alternatif apabila kondisi bernilai False, seperti yang diterapkan pada validasi kelogisan jumlah tiket dalam studi kasus. Sementara itu, struktur if-else menyediakan dua jalur eksekusi yang saling eksklusif, di mana salah satu blok pasti akan dijalankan tergantung hasil evaluasi kondisi, sebagaimana diterapkan pada evaluasi kelayakan umur pengunjung wahana non-ATV dan penentuan bonus souvenir berdasarkan total pembayaran. Lebih lanjut, struktur if-elif-else yang dikombinasikan dengan operator logika (and, !=) memungkinkan pengujian beberapa kondisi majemuk sekaligus dalam satu rantai pemilihan tanpa perlu menyarangkan (nesting) satu if di dalam if lain, sehingga kode tetap ringkas dan mudah dibaca, seperti pada validasi izin kendali wahana Motor ATV Ekstrim yang mempertimbangkan tiga variabel sekaligus, yaitu paket, umur, dan kepemilikan SIM.

Selain itu, praktikum ini juga menunjukkan bahwa Multi-IF, yakni rangkaian if yang berdiri sendiri-sendiri tanpa elif maupun else, berbeda secara fundamental dari if-elif-else, karena setiap kondisi dievaluasi secara independen sehingga memungkinkan lebih dari satu blok kode dieksekusi sekaligus. Konsep ini terbukti esensial dalam menghitung akumulasi diskon bertingkat, seperti diskon belanja besar, member, promo, dan rombongan, yang dapat ditumpuk menjadi satu total persentase diskon. Di sisi lain, struktur match-case sebagai fitur pencocokan pola terstruktur (structural pattern matching) yang diperkenalkan sejak Python 3.10 terbukti mampu menggantikan rantai if-elif yang panjang dengan penulisan yang lebih rapi dan terstruktur, khususnya untuk pemilihan di antara banyak nilai diskrit seperti pemilihan nomor paket wahana pada studi kasus, lengkap dengan pola wildcard (case _) sebagai penanganan nilai yang tidak valid.
