## 2.1 Dasar Teori

Operator adalah simbol khusus dalam Python yang digunakan untuk menjalankan suatu operasi terhadap satu atau lebih nilai (operand), sehingga menghasilkan sebuah nilai baru (Jago Ngoding, 2021). Python menyediakan tujuh kelompok operator: aritmatika, penugasan, perbandingan, logika, keanggotaan, identitas, dan bitwise (RevoU, 2024). Berikut penjelasan rinci masing-masing:

### 2.1.1 Perlungan For
Digunakan untuk operasi matematika dasar (Ediweb.dev, 2024). Terdiri dari + (penjumlahan), - (pengurangan), * (perkalian), / (pembagian, hasil selalu pecahan/float), // (pembagian bulat/floor division, hasil dibulatkan ke bawah), % (modulus, mengambil sisa hasil bagi), dan ** (perpangkatan).

### 2.1.2 Nested For
Digunakan untuk memberikan atau memperbarui nilai suatu variabel (Jago Ngoding, 2021). Bentuk paling dasar adalah =. Selain itu tersedia bentuk augmented assignment yang menggabungkan operasi aritmatika sekaligus penugasan ulang, yaitu +=, -=, *=, /=, %=, //=, dan **=.

### 2.1.3 Operator Perbandingan.
Digunakan untuk membandingkan dua operand dan selalu menghasilkan nilai boolean, True atau False (DQLab, 2021). Terdiri dari == (sama dengan), != (tidak sama dengan), > (lebih besar), < (lebih kecil), >= (lebih besar atau sama dengan), dan <= (lebih kecil atau sama dengan).

### 2.1.4 Operator Logika.
Digunakan untuk menggabungkan atau membalik beberapa ekspresi boolean, dan sangat berkaitan erat dengan operator perbandingan karena sama-sama menghasilkan nilai boolean (Jago Ngoding, 2021). Terdiri dari and (bernilai True hanya jika kedua kondisi benar), or (bernilai True jika salah satu kondisi benar), dan not (membalik nilai boolean suatu kondisi).

### 2.1.5 Operator Keanggotaan.
Digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah struktur data seperti list, tuple, atau string. Terdiri dari in (bernilai True jika nilai ditemukan) dan not in (bernilai True jika nilai tidak ditemukan).

### 2.1.6 Operator Identitas.
Digunakan untuk membandingkan identitas dua objek, yaitu apakah keduanya menunjuk ke lokasi memori yang sama persis, bukan sekadar nilainya sama. Terdiri dari is dan is not. Berbeda dengan == yang membandingkan isi/nilai, is membandingkan alamat objek di memori. Dua variabel list dengan isi identik namun dibuat secara terpisah akan menghasilkan is bernilai False, meskipun == tetap bernilai True.

### 2.1.7 Operator Bitwise.
Digunakan untuk mengolah data dalam bentuk representasi biner (bit demi bit), dan biasa dipakai untuk membangun rangkaian logika seperti AND, OR, atau XOR pada tingkat bit, misalnya untuk keperluan flag atau kode status suatu sistem (Duniailkom, 2019; Anbidev, 2019). Operator ini meliputi & (AND, hasil bit bernilai 1 hanya jika kedua bit bernilai 1), | (OR, hasil bit bernilai 1 jika salah satu bit bernilai 1), ^ (XOR, hasil bit bernilai 1 jika kedua bit berbeda), ~ (NOT, membalik seluruh bit), << (shift kiri, menggeser bit ke kiri sejumlah posisi tertentu, setara mengalikan dengan pangkat dua), dan >> (shift kanan, menggeser bit ke kanan, setara membagi dengan pangkat dua).

### 2.2.3 Program Perulangan For 3 "A_2611531007"

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_1007 = int(input("Masukkan jumlah perulangan: "))

jumlah = 0
for i in range(1, ulang_1007 + 1):
    print(i, end=" ")
    jumlah = jumlah + i
    
    if i < ulang_1007:
        print(" + ", end="")
    else:
        print(" = ", jumlah, end="")
print()
print("Jumlah =", jumlah)
```

### 2.2.4 Program Nested For 1 "A_2611531007"

```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_1007 = int(input("Masukkan nilai batas: "))
for line in range(1, batas_1007 + 1):
    for j in range(1, (-1 * line + batas_1007) + 1):
        print(".", end="")
    print(line)
```


Penjelasan Program:

Menggunakan rumus matematika (-1 * line + batas_1007) + 1 menyebabkan jumlah titik yang dicetak akan semakin berkurang pada setiap baris yang baru.
5. Baris 9 Berfungsi mencetak karakter titik . secara menyamping di baris yang sama. Perintah ini akan dijalankan berulang-ulang oleh perulangan dalam (baris 8) sampai jumlahnya sesuai dengan rumus.

### 2.2.6 Program Nested For 4 "A_2611531007"

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_1007 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1007 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_1007 = tinggi_1007
    c_1007 = a_1007
    lebar_1007 = (2 * tinggi_1007) - 2

    for i_1007 in range(1, tinggi_1007 + 1):
        b_1007 = c_1007 + 1
        
        for j_1007 in range(1, lebar_1007 + 1):
            
            # Baris atas dan bawah
            if i_1007 == 1 or i_1007 == tinggi_1007:
                if j_1007 == 1 or j_1007 == lebar_1007:
                    print("#", end="")
                else:
                    print("=", end="")
            # Baris isi
            else:
                if j_1007 == 1 or j_1007 == lebar_1007:
                    print("|", end="")
                else:
                    if j_1007 == c_1007:
                        print("<", end="")
                    elif j_1007 == b_1007:
                        print(">", end="")
                    elif j_1007 == (lebar_1007 - c_1007):
                        print("<", end="")
                    elif j_1007 == (lebar_1007 - c_1007 + 1):
                        print(">", end="")
                    elif j_1007 > b_1007 and j_1007 < (lebar_1007 - c_1007):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_1007 -= 2

        if a_1007 <= 0:
            c_1007 = (-a_1007) + 2
        else:
            c_1007 = a_1007
```

Penjelasan Program:

1. Baris 1,2,3, dan 4 adalah komentar (comment)
2. Baris 6 Berfungsi meminta pengguna memasukkan angka (diharapkan angka genap), mengubahnya menjadi bilangan bulat (int), lalu menyimpannya ke dalam variabel tinggi_1007.
3. Baris 8 Berfungsi melakukan pengecekan bersyarat (percabangan). Operator % (modulo) digunakan untuk mencari sisa bagi. Jika angka dibagi 2 sisanya tidak sama dengan (!=) 0, maka itu adalah bilangan ganjil.
4. Baris 9 Jika kondisi di baris 8 benar (inputnya ganjil), maka program akan mencetak pesan peringatan ini ke layar.
5. Baris 10 Jika kondisi di baris 8 salah (artinya input yang dimasukkan sudah benar berupa bilangan genap), maka program akan mengeksekusi blok kode di bawah ini.
6. Baris 11 Berfungsi membuat variabel a_1007 dan mengisinya dengan nilai dari input tinggi_1007.
7. Baris 12 Berfungsi membuat variabel c_1007 dan mengisinya dengan nilai dari a_1007.
8. Baris 13 Berfungsi menghitung total lebar dari pola yang akan dibuat menggunakan rumus (2 * tinggi) - 2, lalu menyimpannya di variabel lebar_1007.

```python
print("===== SISTEM TRANSAKSI TOKO =====\n")

# 1. INPUT DATA PELANGGAN DAN TRANSAKSI
nama_1007 = input("Masukkan Nama Pelanggan : ")
status_1007 = input("Masukkan Status Pelanggan (member/nonmember) : ")
total_1007 = float(input("Masukkan Total Belanja : "))
jumlah_1007 = int(input("Masukkan Jumlah Barang : "))
promo_1007 = input("Masukkan Kode Promo : ")

print("\n===== DATA TRANSAKSI =====")
print(f"Nama Pelanggan       : {nama_1007}")
print(f"Status Pelanggan     : {status_1007}")
print(f"Total Belanja        : Rp{total_1007:}")
print(f"Jumlah Barang        : {jumlah_1007}")
print(f"Kode Promo           : {promo_1007}")

# Daftar kode promo yang tersedia di toko
daftar_promo_1007 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]

# 2. OPERATOR PERBANDINGAN
syarat_belanja_1007 = total_1007 >= 200000          
syarat_barang_1007 = jumlah_1007 >= 3                
is_member_1007 = status_1007 == "member"             

# 3. OPERATOR KEANGGOTAAN (in / not in)
promo_tersedia_1007 = promo_1007 in daftar_promo_1007
promo_tidak_tersedia_1007 = promo_1007 not in daftar_promo_1007


# 4. OPERATOR LOGIKA (and, or, not)
diskon_member_1007 = is_member_1007 and syarat_belanja_1007
dapat_promo_1007 = promo_tersedia_1007 and (syarat_belanja_1007 or syarat_barang_1007)
bukan_member_1007 = not is_member_1007

print("\n===== HASIL VALIDASI =====")
print(f"Belanja >= Rp200000        : {syarat_belanja_1007}")
print(f"Jumlah Barang >= 3         : {syarat_barang_1007}")
print(f"Status Member              : {is_member_1007}")
print(f"Kode Promo Tersedia        : {promo_tersedia_1007}")
print(f"Mendapatkan Diskon Member  : {diskon_member_1007}")
print(f"Mendapatkan Promo          : {dapat_promo_1007}")

# 5. OPERATOR ARITMATIKA
diskon_persen_1007 = 0.10 if diskon_member_1007 else 0.0
besar_diskon_1007 = total_1007 * diskon_persen_1007
total_bayar_1007 = total_1007 - besar_diskon_1007
rata_rata_1007 = total_bayar_1007 / jumlah_1007

# 6. OPERATOR PENUGASAN (termasuk augmented assignment)
akumulasi_bayar_1007 = 0
akumulasi_bayar_1007 += total_bayar_1007
akumulasi_bayar_1007 *= 1

print("\n===== HASIL PERHITUNGAN =====")
print(f"Diskon                     : Rp{besar_diskon_1007:.0f}")
print(f"Total Pembayaran           : Rp{total_bayar_1007:.0f}")
print(f"Rata-rata Harga Barang     : Rp{rata_rata_1007:.0f}")

# 7. OPERATOR IDENTITAS (is / is not)
obj_a_1007 = daftar_promo_1007
obj_b_1007 = daftar_promo_1007
obj_c_1007 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]

identitas_sama_1007 = obj_a_1007 is obj_b_1007
identitas_beda_1007 = obj_a_1007 is not obj_c_1007
nilai_sama_1007 = obj_a_1007 == obj_c_1007

# 8. OPERATOR BITWISE - Kode Status Transaksi
bit_member_1007 = 0b0001 if is_member_1007 else 0b0000
bit_belanja_1007 = 0b0010 if syarat_belanja_1007 else 0b0000
bit_barang_1007 = 0b0100 if syarat_barang_1007 else 0b0000
bit_promo_1007 = 0b1000 if promo_tersedia_1007 else 0b0000

# OR (|) menggabungkan seluruh kondisi menjadi satu kode status
kode_status_1007 = bit_member_1007 | bit_belanja_1007 | bit_barang_1007 | bit_promo_1007

# AND (&) memeriksa apakah kondisi tertentu aktif pada kode status
cek_member_1007 = kode_status_1007 & 0b0001
cek_promo_1007 = kode_status_1007 & 0b1000


# XOR (^) membandingkan perbedaan status terhadap kode referensi
kode_referensi_1007 = 0b1011
xor_status_1007 = kode_status_1007 ^ kode_referensi_1007

# Bitwise shift sebagai tambahan (bukan wajib, pelengkap)
shift_status_1007 = kode_status_1007 << 1

# Hak akses pelanggan berdasarkan bit
member_access_1007 = bool(cek_member_1007)
promo_access_1007 = bool(cek_promo_1007)
free_shipping_access_1007 = promo_1007 == "GRATISONGKIR"

print("\n===== HAK AKSES PELANGGAN =====")
print(f"Kode Status (biner)        : {bin(kode_status_1007)}")
print(f"Kode Status (desimal)      : {kode_status_1007}")
print(f"Member Access              : {member_access_1007}")
print(f"Promo Access               : {promo_access_1007}")
print(f"Free Shipping Access       : {free_shipping_access_1007}")

#=====OPERASI BITWISE=====
print("\n=== OPERASI BITWISE =====")

print("\n===== Kode Status Transaksi =====")

#KODE STATUS
#0001 = member
#0010 = total belanja >= Rp 200000
#0100 = jumlah barang >= 3
#1000 = promo

#menggunakan OR (|)
kode_transaksi_1007 =  int(is_member_1007) << 0 | int(syarat_belanja_1007) << 1 | int(syarat_barang_1007) << 2 | int(promo_tersedia_1007) << 3
kode_referensi_1007 = int(is_member_1007) << 0 | int(syarat_belanja_1007) << 1 | int(promo_tersedia_1007) << 3

print(f"{format(int(is_member_1007) << 0, '04b')} | {format(int(syarat_belanja_1007) << 1, '04b')} | {format(int(syarat_barang_1007) << 2, '04b')} | {format(int(promo_tersedia_1007) << 3, '04b')}")
print(f"Kode Biner   : {format(kode_transaksi_1007, '04b')}") 
print(f"Kode Desimal : {kode_transaksi_1007}")


#=====PEMERIKSAAN STATUS=====
print("\n===== Pemeriksaan Status =====")


print("\nCek Member")

#menggunakan and (&)

print(f"{format(kode_transaksi_1007, '04b')} & {format(int(is_member_1007) << 0, '04b')}")
print(f"Hasil Biner   : {format((kode_transaksi_1007) & int(is_member_1007) << 0, '04b')}")
print(f"Hasil Desimal : {(kode_referensi_1007) & int(is_member_1007) << 0}") 

print("\nCek Promo")

print(f"{format(kode_transaksi_1007, '04b')} & {format(int(promo_tersedia_1007) << 3, '04b')}")
print(f"Hasil Biner   : {format((kode_transaksi_1007) & int(promo_tersedia_1007) << 3, '04b')}")
print(f"Hasil Desimal  : {(kode_transaksi_1007) & int(promo_tersedia_1007) << 3}")


#=====PERBANDINGAN STATUS=====
print("\n===== Perbandingan Status =====")

 # menggunakan XOR (^)

print(f"Kode Transaksi : {format(kode_transaksi_1007, '04b')}")
print(f"Kode Referensi : {format(kode_referensi_1007, '04b')}")
print(f"{format(kode_transaksi_1007, '04b')} ^ {format(kode_referensi_1007, '04b')}")
print(f"Hasil Biner    : {format((kode_transaksi_1007) ^ (kode_referensi_1007))}")
print(f"Hasil Desimal  : {(kode_transaksi_1007) ^ (kode_referensi_1007)}")


#======SHIFT GESER KIRI=====
print("\n===== Shift =====")

#bitwise kiri/ shift kiri (<<)

print(f"{format(kode_transaksi_1007,'04b')} << 1")
print(f"Hasil Biner   : {format((kode_transaksi_1007) << 1,'04b')}")
print(f"Hasil Desimal : {(kode_transaksi_1007) << 1}") 

print("\n===== SELESAI =====")
```
