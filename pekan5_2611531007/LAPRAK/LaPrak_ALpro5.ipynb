## 2.1 Dasar Teori

Perulangan (looping) adalah struktur kontrol yang membuat program menjalankan sekumpulan instruksi berulang kali tanpa harus menuliskannya satu per satu. Python menyediakan tiga cara untuk melakukannya, yaitu perulangan while, perulangan for, dan rekursi.: 

### 2.1.1 Perulungan For
Perulangan for mengulang blok instruksi untuk setiap elemen dalam suatu urutan (sequence) atau objek yang dapat diiterasi (iterable). Jumlah pengulangannya ditentukan oleh banyaknya elemen, bukan oleh kondisi seperti pada while. Sintaks dasarnya adalah kata kunci for, variabel perulangan, kata kunci in, dan iterable, lalu diakhiri tanda titik dua (:). Instruksi yang diulang ditulis dengan indentasi di bawah baris for. Python memakai indentasi untuk menentukan isi blok perulangan, sehingga instruksi yang menjorok lebih jauh dari baris for dianggap berada di dalam loop.

### 2.1.2 Perulangan Bersarang (Nested For)
Perulangan bersarang adalah perulangan for yang berada di dalam perulangan for lain. Perulangan yang di luar disebut outer loop, dan yang di dalam disebut inner loop. Untuk setiap satu putaran outer loop, inner loop dijalankan penuh dari awal sampai selesai, lalu outer loop maju ke elemen berikutnya. Jika outer loop berputar sebanyak m kali dan inner loop sebanyak n kali, maka instruksi di dalam inner loop dijalankan sebanyak m × n kali. Parameter end=" " membuat print() tidak pindah baris, sedangkan print() kosong di akhir outer loop memindahkan kursor ke baris berikutnya. Perulangan bersarang banyak dipakai untuk membuat tabel perkalian, mencetak pola karakter, mengolah data dua dimensi seperti matriks, dan mendaftar seluruh pasangan kemungkinan dari dua kumpulan data. Lockwood dan De Chenne (2019) memakainya untuk mencetak seluruh pasangan terurut dalam konteks kombinatorik. Pada pencetakan pola, parameter end=" " membuat print() tidak pindah baris, sedangkan print() kosong di akhir outer loop memindahkan kursor ke baris berikutnya.

### 2.2.3 Program Perulangan For 3 "A_2611531007"

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_1007 = int(input("Masukkan jumlah perulangan: "))

jumlah = 0
for i in range(1, ulang_1007 + 1):
    print(i, end=" ")
    jumlah = jumlah + i
    
    if i < ulang_1007:
        print(" + ", end="")
    else:
        print(" = ", jumlah, end="")
print()
print("Jumlah =", jumlah)
```

### 2.2.4 Program Nested For 1 "A_2611531007"

```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_1007 = int(input("Masukkan nilai batas: "))
for line in range(1, batas_1007 + 1):
    for j in range(1, (-1 * line + batas_1007) + 1):
        print(".", end="")
    print(line)
```


Penjelasan Program:

Menggunakan rumus matematika (-1 * line + batas_1007) + 1 menyebabkan jumlah titik yang dicetak akan semakin berkurang pada setiap baris yang baru.

### 2.2.7 Program Nested For 4 "A_2611531007"

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_1007 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1007 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_1007 = tinggi_1007
    c_1007 = a_1007
    lebar_1007 = (2 * tinggi_1007) - 2

    for i_1007 in range(1, tinggi_1007 + 1):
        b_1007 = c_1007 + 1
        
        for j_1007 in range(1, lebar_1007 + 1):
            
            # Baris atas dan bawah
            if i_1007 == 1 or i_1007 == tinggi_1007:
                if j_1007 == 1 or j_1007 == lebar_1007:
                    print("#", end="")
                else:
                    print("=", end="")
            # Baris isi
            else:
                if j_1007 == 1 or j_1007 == lebar_1007:
                    print("|", end="")
                else:
                    if j_1007 == c_1007:
                        print("<", end="")
                    elif j_1007 == b_1007:
                        print(">", end="")
                    elif j_1007 == (lebar_1007 - c_1007):
                        print("<", end="")
                    elif j_1007 == (lebar_1007 - c_1007 + 1):
                        print(">", end="")
                    elif j_1007 > b_1007 and j_1007 < (lebar_1007 - c_1007):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_1007 -= 2

        if a_1007 <= 0:
            c_1007 = (-a_1007) + 2
        else:
            c_1007 = a_1007
```

# <h1 align = "center">KESIMPULAN <h1> 

Perulangan for pada Python adalah struktur kontrol yang digunakan untuk mengulang blok instruksi pada setiap elemen dalam suatu iterable, seperti range(), list, tuple, string, dan dictionary. Jumlah pengulangannya ditentukan oleh banyaknya elemen yang diiterasi. Sintaksnya terdiri dari kata kunci for, variabel perulangan, kata kunci in, dan iterable, lalu diakhiri tanda titik dua (:). Instruksi yang diulang harus ditulis dengan indentasi, karena Python memakai indentasi untuk menentukan isi blok perulangan. Untuk pengulangan dengan jumlah tertentu, fungsi range() dapat dipakai dengan mengatur parameter start, stop, dan step, dengan catatan nilai stop tidak ikut diproses. Alur perulangan juga dapat dikendalikan menggunakan pernyataan break, continue, dan else sesuai kebutuhan program.

Perulangan bersarang (nested for), yaitu perulangan for di dalam perulangan for lain. Pada perulangan ini, setiap satu putaran outer loop menjalankan inner loop sampai selesai, sehingga jumlah eksekusi instruksi di dalamnya adalah hasil kali jumlah putaran kedua loop. Perulangan bersarang berguna untuk mencetak pola, mengolah data dua dimensi seperti matriks, dan membuat tabel. Kesalahan yang paling sering terjadi dalam penulisan perulangan for adalah indentasi yang salah, lupa tanda titik dua, salah memahami batas range(), dan menerapkan for pada objek yang bukan iterable.
